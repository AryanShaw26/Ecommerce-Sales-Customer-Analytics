```python
reference_date = df_customer["InvoiceDate"].max()

rfm = df_customer.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (reference_date - x.max()).days),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("Revenue", "sum")
)

rfm.head()
```

```python
rfm.describe()
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head()
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

rfm.head()
```

```python
def segment_customer(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "Champions"
    elif row["R_Score"] >= 4 and row["F_Score"] >= 3:
        return "Loyal Customers"
    elif row["R_Score"] >= 4 and row["M_Score"] >= 4:
        return "Potential Loyalists"
    elif row["R_Score"] >= 3 and row["F_Score"] <= 2:
        return "New Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 4:
        return "At Risk"
    elif row["R_Score"] <= 2 and row["M_Score"] >= 4:
        return "High Value At Risk"
    elif row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Lost Customers"
    else:
        return "Needs Attention"


rfm["Segment"] = rfm.apply(segment_customer, axis=1)

rfm.head()
```

```python
segment_summary = (
    rfm.groupby("Segment")
    .agg(
        Customers=("Segment", "count"),
        Revenue=("Monetary", "sum"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean")
    )
    .sort_values("Revenue", ascending=False)
)

segment_summary
```

```python
product_sales = df_sales[
    (df_sales["TransactionType"] == "Sale") &
    (df_sales["StockCode"] != "DOT")
].copy()

product_sales_summary = (
    product_sales
    .groupby(["StockCode", "Description"])
    .agg(
        Revenue=("Revenue", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("InvoiceNo", "nunique")
    )
)

product_sales_summary["Revenue_per_Unit"] = (
    product_sales_summary["Revenue"] /
    product_sales_summary["Quantity"]
)

product_sales_summary = product_sales_summary.sort_values(
    "Revenue",
    ascending=False
)

product_sales_summary.head(10)
```

```python
return_data = df_sales[
    df_sales["TransactionType"] == "Return"
].copy()

return_revenue = return_data["Revenue"].sum()
return_quantity = return_data["Quantity"].sum()
return_orders = return_data["InvoiceNo"].nunique()

return_revenue, return_quantity, return_orders
```

```python
sales_data = df_sales[
    df_sales["TransactionType"] == "Sale"
]

sales_revenue = sales_data["Revenue"].sum()
sales_quantity = sales_data["Quantity"].sum()
sales_orders = sales_data["InvoiceNo"].nunique()

return_rate_revenue = abs(return_revenue) / sales_revenue * 100
return_rate_quantity = abs(return_quantity) / sales_quantity * 100
return_rate_orders = return_orders / sales_orders * 100

return_rate_revenue, return_rate_quantity, return_rate_orders
```

```python
product_returns = (
    return_data[return_data["StockCode"] != "DOT"]
    .groupby(["StockCode", "Description"])
    .agg(
        Returned_Quantity=("Quantity", "sum"),
        Returned_Revenue=("Revenue", "sum"),
        Return_Orders=("InvoiceNo", "nunique")
    )
)

product_returns["Returned_Quantity"] = (
    product_returns["Returned_Quantity"].abs()
)

product_returns["Returned_Revenue"] = (
    product_returns["Returned_Revenue"].abs()
)

product_returns.sort_values(
    "Returned_Quantity",
    ascending=False
).head(10)
```

```python
product_return_analysis = product_sales_summary[
    ["Revenue", "Quantity", "Orders"]
].join(
    product_returns[
        ["Returned_Quantity", "Returned_Revenue", "Return_Orders"]
    ],
    how="left"
)

product_return_analysis = product_return_analysis.fillna(0)

product_return_analysis["Return_Rate_%"] = (
    product_return_analysis["Returned_Quantity"]
    / product_return_analysis["Quantity"]
    * 100
)

product_return_analysis.sort_values(
    "Return_Rate_%",
    ascending=False
).head(10)
```

```python
product_return_rate = product_return_analysis[
    product_return_analysis["Quantity"] >= 100
].copy()

product_return_rate = product_return_rate.sort_values(
    "Return_Rate_%",
    ascending=False
)

product_return_rate.head(10)
```

```python
product_return_rate_filtered = product_return_analysis[
    product_return_analysis["Orders"] >= 20
].copy()

product_return_rate_filtered = (
    product_return_rate_filtered
    .sort_values("Return_Rate_%", ascending=False)
)

product_return_rate_filtered.head(10)
```
